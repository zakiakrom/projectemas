```python
import pandas as pd

path = r'C:\PROJE\yapayzeka\Notebooks\Ecommerce_Sales_Data_2024_2025 (4).csv'
df = pd.read_csv(path)

print("Data Berhasil Terbaca!")
print(df.head())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Bersihkan nama kolom dulu
df.columns = df.columns.str.strip()

# 2. Cek apakah kolom benar-benar ada setelah dibersihkan
if 'Category' in df.columns:
    plt.figure(figsize=(12, 6))
    # Gunakan palet warna agar lebih menarik
    sns.barplot(data=df, x='Category', y='Profit', estimator=sum, palette='magma')
    plt.title('Total Profit per Kategori')
    plt.xticks(rotation=45)
    plt.show()
else:
    print(f"Kolom 'Category' tidak ditemukan. Kolom yang tersedia adalah: {df.columns.tolist()}")
```
