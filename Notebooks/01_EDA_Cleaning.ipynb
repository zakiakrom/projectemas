# Step 1: Data Cleaning & Preprocessing

Tahap ini bertujuan untuk mengubah data mentah yang berantakan menjadi data "bersih" yang siap diproses oleh algoritma AI.

```python
import pandas as pd
import numpy as np

# 1. Load Dataset
file_path = r'D:\finish\New folder\Kuliah\Semester 6\Yapay Zeka\yapayzeka\Data\Processed\Ecommerce_Sales_Data_2024_2025 (4).csv'
df = pd.read_csv(file_path)

# 2. Standardize column names (lowercase & underscores)
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_').str.replace('-', '_')

# 3. Fix date format
if 'order_date' in df.columns:
    df['order_date'] = pd.to_datetime(df['order_date'])

# 4. Remove duplicates and fill missing values with Median
df = df.drop_duplicates()
numeric_cols = df.select_dtypes(include=[np.number]).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())

# 5. Create 'is_trending' target based on quantity median
if 'quantity' in df.columns:
    median_qty = df['quantity'].median()
    df['is_trending'] = (df['quantity'] > median_qty).astype(int)

# 6. Save cleaned data
output_path = r'D:\finish\New folder\Kuliah\Semester 6\Yapay Zeka\yapayzeka\Data\Processed\ecommerce_clean.csv'
df.to_csv(output_path, index=False)

# 7. Final Check
print(f"Total Rows: {len(df)}")
print(f"Columns: {df.columns.tolist()}")
df.head()
```
